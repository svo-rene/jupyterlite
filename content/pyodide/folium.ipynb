# Interaktive Karten mit `folium`

Eine einfache Demonstration, wie sich eine Karte in einem `jupyterlite`-Notebook darstellen lässt.

Beachte: Das Paket `folium` hat mehrere Abhängigkeiten, die ihrerseits wieder Abhängigkeiten haben können.

Der folgende Codeausschnitt zeigt – ausgeführt in einer frischen Python-Umgebung, in der `folium` bereits installiert ist –, welche Pakete beim Import von `folium` mitgeladen werden:

```python
#https://stackoverflow.com/a/40381601/454773
import sys
before = [str(m) for m in sys.modules]
import folium
after = [str(m) for m in sys.modules]
set([m.split('.')[0] for m in after if not m in before and not m.startswith('_')])
```
