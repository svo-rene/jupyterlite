# p5-Notebook

Eine minimale Jupyter-Notebook-Oberfläche für [p5.js](https://p5js.org)-Kernel.

Zuerst definieren wir ein paar Variablen:

```p5js
var n = 4;
var speed = 1;
```

## Die Funktion `setup`

Die übliche p5-Funktion `setup`, die die Zeichenfläche (Canvas) anlegt.

```p5js
function setup () {
  createCanvas(innerWidth, innerHeight);
  rectMode(CENTER);
}
```

## Die Funktion `draw`

Aus der [p5.js-Dokumentation](https://p5js.org/reference/#/p5/draw):

> Die Funktion `draw()` führt die Zeilen in ihrem Block fortlaufend aus, bis das Programm gestoppt wird oder `noLoop()` aufgerufen wird.

```p5js
function draw() {
  background('#ddd');
  translate(innerWidth / 2, innerHeight / 2);
  for (let i = 0; i < n; i++) {
    push();
    rotate(frameCount * speed / 1000 * (i + 1));
    fill(i * 5, i * 100, i * 150);
    const s = 200 - i * 10;
    rect(0, 0, s, s);
    pop();
  }
}
```

## Den Sketch anzeigen

Jetzt zeigen wir den Sketch mit der Magic `%show` an:

```p5js
%show
```

```p5js
n = 20
```
